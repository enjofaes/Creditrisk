# Probability of Default Estimation

```python
from scipy.stats import norm
import numpy as np

def distance_to_default(V, D, sigma, T=1):
    dd = (np.log(V/D) + 0.5*sigma**2*T)/(sigma*T**0.5)
    return dd, norm.cdf(-dd)

dd, pd = distance_to_default(120e6, 80e6, 0.25)
print(f"DD={dd:.2f}, PD={pd:.2%}")
```
