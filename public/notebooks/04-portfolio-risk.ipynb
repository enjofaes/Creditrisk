# Portfolio Credit Risk

```python
import numpy as np

np.random.seed(42)
n_sims, n_loans, pd, lgd, rho = 10000, 100, 0.03, 0.45, 0.15

Z = np.random.standard_normal(n_sims)
eps = np.random.standard_normal((n_sims, n_loans))
from scipy.stats import norm
R = np.sqrt(rho)*Z[:,None] + np.sqrt(1-rho)*eps
losses = (R < norm.ppf(pd)).sum(axis=1) * lgd * 10000
print(f"EL={losses.mean():,.0f}, VaR99={np.percentile(losses,99):,.0f}")
```
