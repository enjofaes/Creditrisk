# Neural Networks for Credit

```python
from sklearn.neural_network import MLPClassifier
import numpy as np
from sklearn.metrics import roc_auc_score

np.random.seed(0)
X = np.random.randn(2000, 4)
y = (X[:,0] - X[:,1] > 0.5).astype(int)
model = MLPClassifier(hidden_layer_sizes=(32,16), max_iter=200).fit(X, y)
auc = roc_auc_score(y, model.predict_proba(X)[:,1])
print(f"AUC: {auc:.3f}")
```
