# WoE and IV Analysis

```python
import numpy as np, pandas as pd

np.random.seed(0)
score = np.random.normal(680, 80, 5000).clip(300,850)
y = (score < 620).astype(int)
df = pd.DataFrame({"score": score, "default": y})
df["bin"] = pd.qcut(df["score"], 8, duplicates="drop")
print(df.groupby("bin")["default"].mean().to_string())
```
